Array Attributes


```python
a1 = np.arange(10)
a2 = np.arange(12, dtype = float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)
```

```python
#ndim
print(a1.ndim)
print(a2.ndim)
print(a3.ndim)
```

```python
#shape
print(a1.shape)
print(a2.shape)
print(a3.shape)
```

```python
#size
print(a1.size)
print(a2.size)
print(a3.size)
```

```python
#itemsize
print(a1.itemsize)
print(a2.itemsize)
print(a3.itemsize)
```

Changing Datatypes

```python
#astype
a1.astype(np.float32)
a3.astype(np.int32)
```

Array Operations

```python
a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(3,4)

a2
```

```python
#scalar operations
#arithmetic
a1 * 2
a2 ** 2
```

```python
#vector operations
#arithmetic
a1 + a2
```
